```python
import pandas as pd
df = pd.read_csv('data.csv')
```

```python
display(df.head())
```

```python
print("информация о данных")
df.info()
```

```python
print("Количество пропусков")
print(df.isnull().sum())
```

```python
print("Статистика по колонке Price")
df['Price'].describe()
```

```python
print("Размах")
price_range = df['Price'].max() - df['Price'].min()
print(price_range)
```

```python
print("Статистика по колонке Value")
df['Value'].describe()
```

```python
print("Размах")
Value_range = df['Value'].max() - df['Value'].min()
print(Value_range)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12,5))

axes[0].hist(df['Price'], bins=8, edgecolor='black', alpha=0.7)
axes[0].set_title('Распределение Price')
axes[0].set_xlabel('Цена')
axes[0].set_ylabel('Частота')

axes[1].boxplot(df['Price'])
axes[1].set_title('Ящик с усами для Price')
axes[1].set_ylabel('Цена')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(df['Value'], bins=8, edgecolor='black', alpha=0.7)
axes[0].set_title('Распределение Value')
axes[0].set_xlabel('Значение')
axes[0].set_ylabel('Частота')

axes[1].boxplot(df['Value'])
axes[1].set_title('Ящик с усами для Value')
axes[1].set_ylabel('Значение')

plt.tight_layout()
plt.show()
```

Стандартное отклонение цены достаточно большое (7.666743). Это больше половины среднего значения. Товары силльно отличаются по стоимости. 
Среднее и медиана, как у категорий товаров, так и у цены, сильно не отличаются. Это означает отсутствие выбросов. Данные достаточно равномерные. 
Средний ценовой сегмент находится между 9.75 и 18.25
Из-за небольшой выборки гистограммы имеют "ямы"
Изменение цен плавное и предсказуемое. Категории товаров - нет.  Они делятся на две большие группы.
